```python
from smolagents import tool, CodeAgent, InferenceClientModel

#tool to suggest a menu based on the occasion
@tool
def suggest_menu(occasion: str) -> str:
    """
    Return a predefined menu for an occasion.

    Args:
        occasion: One of 'casual', 'formal', or 'superhero'.
            Use 'formal' for formal parties and elegant events.
            Use 'casual' for informal gatherings.
            Use 'superhero' for superhero-themed parties.

    Returns:
        The predefined menu for the occasion.
    """
    if occasion == "casual":
        return "Casual menu: Pizza, Sandwiches, Salad"
    elif occasion == "formal":
        return "Formal menu: Champagne, Canapés, Roast Chicken, Salad"
    elif occasion == "superhero":
        return "Superhero menu: Chips, Salsa, Burgers, Hot Dogs"
    else:
        return "Menu not available for this occasion."

model = OpenAIServerModel(
    model_id="qwen2:7b",
    api_base="http://localhost:11434/v1",
    api_key="ollama"
)

agent = CodeAgent(
    tools=[suggest_menu],
    model=model,
    code_block_tags="markdown",
    max_steps=5,
)

result = agent.run(
    "Prepare a formal menu for the party"
)

print(result)
```

```python
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo
from smolagents import CodeAgent, OpenAIServerModel, ToolCallingAgent, tool


@tool
def calculate_party_time(
    drinks: int,
    food: int,
    menu: int,
    music: int
) -> str:
    """
    Calculate when a party will be ready in IST.

    Args:
        drinks: Minutes required to prepare drinks.
        food: Minutes required to prepare food.
        menu: Minutes required to prepare the menu.
        music: Minutes required to prepare music.

    Returns:
        The total preparation time and party-ready time in IST.
    """
    now = datetime.now(ZoneInfo("Asia/Kolkata"))

    total = drinks + food + menu + music

    ready_at = now + timedelta(minutes=total)

    return (
        f"Total preparation time: {total} minutes. "
        f"Party ready at: "
        f"{ready_at.strftime('%d %B %Y, %I:%M %p IST')}"
    )

model = OpenAIServerModel(
    model_id="qwen2:7b",
    api_base="http://localhost:11434/v1",
    api_key="ollama"
)

agent = ToolCallingAgent(
    tools=[calculate_party_time],
    model=model,
    max_steps=3
)

result = agent.run(
    """
    Alfred needs to prepare for a party.

    Drinks: 30 minutes
    Food: 60 minutes
    Menu: 45 minutes
    Music: 45 minutes

    Call calculate_party_time with these durations.
    Return the tool's result without changing it.
    """
)

print(result)
```
